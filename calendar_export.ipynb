# Packages Import

```python
import os
from datetime import datetime 
import pandas as pd
from icalendar import Calendar, Event, vDatetime
import zoneinfo
import uuid
```

# Calendar Export

```python
group_name=input("Enter group name: ")
# ZICSS1-1211
```

```python
classes=pd.read_csv(f"./schedules/{group_name}.csv", encoding="utf-8")
print(classes)
```

```python
if not os.path.exists("calendars"):
    os.mkdir("calendars")
```

```python
# cf=open(f"calendars/{group_name}.ics", "wb")
```

```python
calendar = Calendar()
# calendar["dtstart"] = "20260223000000"
# calendar["dtend"] ="20260930235959"
calendar["summary"] =f"Schedule of the {group_name} group."
calendar["version"] = "2.0"
calendar["PRODID"] = f".//Apollo schedule export//{group_name}"
```

```python
for index, row in classes.iterrows():
    event=Event()
    event["dtstart"] = row["Termin"].replace("-", "")+"T"+row["Start time"].replace(":","")+"00"
    event["dtend"] = row["Termin"].replace("-", "")+"T"+row["End time"].replace(":","") + "00"
    event["location"] = row["Sala"]
    event["summary"] = row["Przedmiot"]
    event["description"] = row["Nauczyciel"]
    event["uid"] = str(uuid.uuid4())
    event["dtstamp"]=vDatetime(datetime.now())
    calendar.add_component(event)
print(calendar.to_ical())
```

```python
with open(f"calendars/{group_name}.ics", "wb") as cf:
    cf.write(calendar.to_ical())
```

```python
cf.close()
```
