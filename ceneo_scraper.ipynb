```python
import requests
from bs4 import BeautifulSoup
```

# Ceneo Scraper 

1. Provide url of the product's opinions page


```python
product_code="135965172"
url= f"https://www.ceneo.pl/{product_code}#tab=reviews"
```

2. Send request to provided url 

```python


r=requests.get(url)
print(r.status_code)
```

3. Fetch product name 

```python
page_dom=BeautifulSoup(r.text, 'html.parser')
print(type(page_dom))
```

```python
product_name=page_dom.select_one('h1').get_text()
print(product_name)
print(type(product_name))
```

# my Idea


```python


if r.status_code==200:
    # print(r.content)
    # soup=BeautifulSoup(r.content,'html.parser')
    reviews=page_dom.find_all('div',{"class":"user-post user-post__card js_product-review"})
    # soup2=BeautifulSoup(reviews,'html.parser')
    for review in reviews:
        print(review.find('div',{'class':"user-post__text"}).get_text())
else:
    print('Error')
```

```python


if r.status_code==200:
    # print(r.content)
    # soup=BeautifulSoup(r.content,'html.parser')
    reviews=page_dom.select_one("div.js_product-reviews.js_reviews-hook.js_product-reviews-container")
    reviews=reviews.select("div.user-post.user-post__card.js_product-review")
    # soup2=BeautifulSoup(reviews,'html.parser')
  
    for review in reviews:
        print(review.select_one("div.user-post__text").get_text())
else:
    print('Error')
```

4. Fetch all opinions from the webpage

```python
opinions=page_dom.find_all('div',class_='js_product-review')
print(type(opinions))
print(len(opinions))
```
